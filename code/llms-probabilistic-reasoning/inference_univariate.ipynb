```python
import numpy as np
import plotly.graph_objects as go
import plotly.io as pio
from scipy import stats

# Parameters for the Gaussian distribution
mu = 170  # mean height in cm
sigma = 10  # standard deviation in cm
threshold = 180  # threshold height

# Create x values for the plot
x = np.linspace(mu - 4*sigma, mu + 4*sigma, 1000)

# Calculate the PDF
pdf = stats.norm.pdf(x, mu, sigma)

# Calculate the probability P(X > 180)
prob_greater = 1 - stats.norm.cdf(threshold, mu, sigma)

# Create the figure
fig = go.Figure()

# Add the full PDF curve
fig.add_trace(go.Scatter(
    x=x,
    y=pdf,
    mode='lines',
    name='PDF',
    line=dict(color='#1f77b4', width=2),
    fill='tozeroy',
    fillcolor='rgba(31, 119, 180, 0.1)'
))

# Add shaded area for P(X > 180)
x_shaded = np.linspace(threshold, mu + 4*sigma, 500)
pdf_shaded = stats.norm.pdf(x_shaded, mu, sigma)
fig.add_trace(go.Scatter(
    x=np.concatenate([[threshold], x_shaded, [mu + 4*sigma], [threshold]]),
    y=np.concatenate([[0], pdf_shaded, [0], [0]]),
    mode='lines',
    name=f'P(X > {threshold})',
    fill='toself',
    fillcolor='rgba(255, 127, 14, 0.5)',
    line=dict(color='rgba(255, 127, 14, 0.5)', width=0),
    showlegend=True
))

# Add vertical line at 180cm
fig.add_trace(go.Scatter(
    x=[threshold, threshold],
    y=[0, stats.norm.pdf(threshold, mu, sigma)],
    mode='lines',
    name=f'x = {threshold}cm',
    line=dict(color='red', width=2, dash='dash'),
    showlegend=True
))

# Add annotation with probability value
fig.add_annotation(
    x=threshold + 5,
    y=stats.norm.pdf(threshold, mu, sigma) * 0.7,
    text=f'P(X > {threshold}) = {prob_greater:.4f}<br>({prob_greater*100:.2f}%)',
    showarrow=True,
    arrowhead=2,
    arrowcolor='red',
    arrowwidth=2,
    bgcolor='white',
    bordercolor='red',
    borderwidth=2,
    font=dict(size=12, color='black')
)

# Update layout
fig.update_layout(
    title=dict(
        text='Univariate Gaussian Distribution: P(X > 180cm)',
        x=0.5,
        font=dict(size=16)
    ),
    xaxis=dict(
        title='Height (cm)',
        range=[mu - 4*sigma, mu + 4*sigma]
    ),
    yaxis=dict(
        title='Probability Density',
        range=[0, max(pdf) * 1.1]
    ),
    width=800,
    height=500,
    template='plotly_white',
    legend=dict(
        x=0.02,
        y=0.98,
        bgcolor='rgba(255, 255, 255, 0.8)',
        bordercolor='black',
        borderwidth=1
    )
)

# Add distribution parameters as text
fig.add_annotation(
    x=0.98,
    y=0.98,
    xref='paper',
    yref='paper',
    text=f'X ~ N(μ = {mu}, σ = {sigma})',
    showarrow=False,
    align='right',
    bgcolor='rgba(255, 255, 255, 0.8)',
    bordercolor='black',
    borderwidth=1,
    font=dict(size=11)
)

fig.show()

# Save the figure as HTML content
html_content = pio.to_html(
    fig,
    include_plotlyjs='cdn',
    full_html=True,
    div_id="univariate-gaussian-plot"
)

print(html_content)
```
